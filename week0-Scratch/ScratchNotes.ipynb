```python
# chat.py
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    input="In one sentence, what is CS50?",
    model="gpt-5"
)

print(response.output_text)
```

output may vary: "CS 50 is Harvard University's introductory computer science course, the intellectual enterprose of Computer Science and the Art of programming."

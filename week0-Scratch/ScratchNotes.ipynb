```python
# chat.py
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    input="In one sentence, what is CS50?",
    model="gpt-5"
)

print(response.output_text)
```

<strong>Output:</strong><br>
CS 50 is Harvard University's introductory computer science course--"the intellectual enterprose of Computer Science and the Art of programming"-- covering problem solving algorithms, data structures, and more using languages like C, Python, and SQL

```python
# Making the code more dynamic

# chat.py
from openai import OpenAI

client = OpenAI()

prompt = input("Prompt: ")

response = client.responses.create(
    input=prompt,
    model="gpt-5"
)

print(response.output_text)
```
